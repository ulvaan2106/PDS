```python

import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

DATASET_PATH = "/content/image_dataset_extracted"
IMAGE_EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
IMG_SIZE = (24, 24)
MAX_IMAGES_PER_CLASS = 1000

X_images = []
y_labels = []
class_names = []

# Find folders that directly contain images
for root, dirs, files in os.walk(DATASET_PATH):
    image_files = [
        f for f in files
        if f.lower().endswith(IMAGE_EXTS)
    ]

    if image_files:
        class_name = os.path.basename(root)

        # Skip the root if all images are mixed together
        class_names.append((class_name, root, image_files))

if not class_names:
    raise ValueError(
        "No images found. Complete extraction and run Cell 2 first."
    )

# Load up to 1000 images from each detected folder
for label, (class_name, folder, files) in enumerate(class_names):
    print("Loading:", class_name)

    for filename in sorted(files)[:MAX_IMAGES_PER_CLASS]:
        path = os.path.join(folder, filename)

        try:
            with Image.open(path) as img:
                img = img.convert("RGB").resize(IMG_SIZE)
                X_images.append(
                    np.asarray(img, dtype=np.float32) / 255.0
                )
                y_labels.append(label)
        except Exception:
            continue

    print("Images loaded:", sum(1 for y in y_labels if y == label))

X_images = np.array(X_images, dtype=np.float32)
y_labels = np.array(y_labels)

print("\nImage array shape:", X_images.shape)
print("Number of labels:", len(y_labels))
print("Detected folders:", [c[0] for c in class_names])

# Display sample images
if len(X_images) > 0:
    fig, axes = plt.subplots(2, 5, figsize=(13, 6))

    for ax in axes.flat:
        idx = np.random.randint(len(X_images))
        ax.imshow(X_images[idx])
        ax.set_title(class_names[y_labels[idx]][0])
        ax.axis("off")

    plt.suptitle("Sample Images from Dataset")
    plt.tight_layout()
    plt.show()
```

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Convert each image into a row of numerical pixel values
X = X_images.reshape(len(X_images), -1)
y = y_labels

# Split into training and testing sets
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train_raw))
print("Testing samples:", len(X_test_raw))

# Scale the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

# Reduce features to speed up model training
n_components = min(50, X_train_scaled.shape[0] - 1,
                   X_train_scaled.shape[1])

pca = PCA(n_components=n_components, random_state=42)

X_train = pca.fit_transform(X_train_scaled)
X_test = pca.transform(X_test_scaled)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print(
    "Variance retained:",
    round(pca.explained_variance_ratio_.sum(), 4)
)
```

```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)

# Get readable class names
labels = np.unique(y)
target_names = [
    class_names[int(i)][0] for i in labels
]

# Store accuracy results for comparison later
results = {}

def evaluate_model(name, model):
    predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)
    results[name] = accuracy

    print(name)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("\nClassification Report:")
    print(classification_report(
        y_test,
        predictions,
        labels=labels,
        target_names=target_names,
        zero_division=0
    ))

    # Confusion matrix graph
    fig, ax = plt.subplots(figsize=(7, 6))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions,
        labels=labels,
        display_labels=target_names,
        cmap="Blues",
        xticks_rotation=45,
        ax=ax,
        colorbar=False
    )

    ax.set_title(name + " - Confusion Matrix")
    plt.tight_layout()
    plt.show()

    return predictions

print("Evaluation function created successfully!")
print("Classes:", target_names)
```

```python

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

# Calculate average brightness for each image
brightness = X_images.mean(axis=(1, 2, 3))

# Use the same train/test split as the classification models
Xb_train_raw, Xb_test_raw, yb_train, yb_test = train_test_split(
    X, brightness,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Scale and reduce dimensions
brightness_scaler = StandardScaler()
Xb_train_scaled = brightness_scaler.fit_transform(Xb_train_raw)
Xb_test_scaled = brightness_scaler.transform(Xb_test_raw)

n_components_reg = min(
    50,
    Xb_train_scaled.shape[0] - 1,
    Xb_train_scaled.shape[1]
)

brightness_pca = PCA(
    n_components=n_components_reg,
    random_state=42
)

Xb_train = brightness_pca.fit_transform(Xb_train_scaled)
Xb_test = brightness_pca.transform(Xb_test_scaled)

# Train model
linear_model = LinearRegression()
linear_model.fit(Xb_train, yb_train)

# Predict brightness
brightness_pred = linear_model.predict(Xb_test)

# Evaluation metrics
print("LINEAR REGRESSION RESULTS")
print("MAE:", round(mean_absolute_error(yb_test, brightness_pred), 4))
print("MSE:", round(mean_squared_error(yb_test, brightness_pred), 4))
print("RMSE:", round(np.sqrt(mean_squared_error(yb_test, brightness_pred)), 4))
print("R2 Score:", round(r2_score(yb_test, brightness_pred), 4))

# Graph 1: Actual vs predicted
plt.figure(figsize=(7, 5))
plt.scatter(yb_test, brightness_pred, alpha=0.6)

low = min(yb_test.min(), brightness_pred.min())
high = max(yb_test.max(), brightness_pred.max())
plt.plot([low, high], [low, high], "r--")

plt.xlabel("Actual Brightness")
plt.ylabel("Predicted Brightness")
plt.title("Linear Regression: Actual vs Predicted")
plt.tight_layout()
plt.show()

# Graph 2: Residual plot
residuals = yb_test - brightness_pred

plt.figure(figsize=(7, 5))
plt.scatter(brightness_pred, residuals, alpha=0.6)
plt.axhline(0, color="red", linestyle="--")

plt.xlabel("Predicted Brightness")
plt.ylabel("Residuals")
plt.title("Linear Regression: Residual Plot")
plt.tight_layout()
plt.show()
```

```python

from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Create and train the Decision Tree
tree_model = DecisionTreeClassifier(
    max_depth=10,
    random_state=42
)

tree_model.fit(X_train, y_train)

# Evaluate the model
tree_predictions = evaluate_model(
    "Decision Tree",
    tree_model
)

# Visualize the first few levels of the tree
plt.figure(figsize=(20, 10))

plot_tree(
    tree_model,
    feature_names=[
        f"PC{i+1}" for i in range(X_train.shape[1])
    ],
    class_names=target_names,
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title("Decision Tree Visualization")
plt.tight_layout()
plt.show()
```

```python

from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import matplotlib.pyplot as plt

# Create and train Random Forest
forest_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

forest_model.fit(X_train, y_train)

# Evaluate the model
forest_predictions = evaluate_model(
    "Random Forest",
    forest_model
)

# Feature importance graph
importance = pd.Series(
    forest_model.feature_importances_,
    index=[
        f"PC{i+1}" for i in range(X_train.shape[1])
    ]
).sort_values(ascending=False).head(15).sort_values()

plt.figure(figsize=(9, 6))
importance.plot(kind="barh")

plt.xlabel("Feature Importance")
plt.ylabel("PCA Feature")
plt.title("Random Forest - Top 15 Important Features")
plt.tight_layout()
plt.show()
```

```python

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Create and train KNN
knn_model = KNeighborsClassifier(
    n_neighbors=5,
    n_jobs=-1
)

knn_model.fit(X_train, y_train)

# Evaluate the model
knn_predictions = evaluate_model(
    "K-Nearest Neighbors (k=5)",
    knn_model
)

# Compare accuracy for different k values
k_values = [1, 3, 5, 7, 9]
knn_accuracies = []

for k in k_values:
    model = KNeighborsClassifier(
        n_neighbors=k,
        n_jobs=-1
    )
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)
    knn_accuracies.append(accuracy)

# Plot accuracy graph
plt.figure(figsize=(8, 5))
plt.plot(
    k_values,
    knn_accuracies,
    marker="o",
    linewidth=2
)

plt.xticks(k_values)
plt.ylim(0, 1.05)
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Test Accuracy")
plt.title("KNN: Accuracy vs Number of Neighbors")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
